The code below keeps only rows where both G and Y are available. In this copy of the table, the columns are given the short names `G` and `Y`, referring to your choices in B. The count table helps you compare data availability across groups.


```python
comparaison = donnees[[groupe, colonne_y]].dropna().copy()
comparaison.columns = ["G", "Y"]
total = donnees[groupe].value_counts().reindex(ordre, fill_value=0)
disponibles = comparaison["G"].value_counts().reindex(ordre, fill_value=0)
print(pd.DataFrame({"total": total, "disponibles": disponibles,
                    "manquantes": total - disponibles}))
bas, haut = comparaison["Y"].min(), comparaison["Y"].max()
marge = max((haut - bas) * 0.05, 0.01)
limites_y = (bas - marge, haut + marge)
```
