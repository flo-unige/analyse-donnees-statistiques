## E : Comment réutiliser une opération sans la réécrire ?

Lorsque nous programmons, certaines tâches reviennent souvent : compter des éléments, arrondir un nombre ou convertir une unité. Une **fonction** permet de regrouper les instructions nécessaires à une tâche et de les réutiliser en lui donnant un nom.

Nous commencerons par des fonctions déjà fournies par Python. Ensuite, nous regarderons comment définir notre propre fonction pour une petite conversion. Le but est surtout de comprendre comment l'utiliser et l'adapter, sans devoir tout retenir par cœur.

### E1 : Demander un résultat à une fonction

Vous pouvez voir un **appel de fonction** comme une demande que vous adressez à un outil. Par exemple, `len(temperatures)` signifie : « compte les éléments de cette liste ». Le nom avant les parenthèses indique l'outil choisi, et les valeurs entre les parenthèses sont les **arguments** que vous lui fournissez.

Ici, `len()` compte les éléments, `max()` retrouve la plus grande valeur et `round()` arrondit un nombre. Dans `round(18.567, ndigits=1)`, le deuxième argument précise que nous souhaitons garder un chiffre après la virgule.

Essayez de prévoir les trois résultats avant d'exécuter. Regardez aussi les deux lignes avec `maximum` : la première conserve le résultat sous ce nom, tandis que la seconde l'affiche.

````python
temperatures = [18.5, 21.0, 19.5]
print(len(temperatures))

maximum = max(temperatures)
print(maximum)

print(round(18.567, ndigits=1))
````

### E2 : Créer une fonction

Supposons que nous ayons souvent besoin de convertir des températures en degrés Celsius vers les kelvins. Nous pouvons définir une fonction qui réalise cette conversion pour nous.

Le mot `def` annonce à Python que nous allons **définir une fonction**. Nous choisissons son nom, `celsius_vers_kelvin`, puis un **paramètre**, `temperature_c`. Ce paramètre est le nom que la fonction utilisera pour désigner la température reçue lors d'un appel.

La ligne commençant par `return` indique le résultat que la fonction doit nous renvoyer. Elle est décalée vers la droite : cette **indentation** montre qu'elle appartient au bloc de la fonction.

Lancez la cellule. Aucun nombre ne s'affiche encore, car nous venons seulement de préparer l'outil. Dans E3, nous l'appellerons en lui donnant une température à convertir.

````python
def celsius_vers_kelvin(temperature_c):
    return temperature_c + 273.15
````

````python
temperature_k = celsius_vers_kelvin(20)
print(temperature_k)
````

## F : Comment donner une logique à notre programme ?

Jusqu'ici, nos instructions se suivaient simplement les unes après les autres. Mais nous avons parfois besoin de choisir une action en fonction d'un résultat, ou de refaire la même opération sur plusieurs mesures.

Nous allons donc découvrir les **conditions**, qui permettent de choisir quelles instructions exécuter, puis les **boucles**, qui permettent de répéter un bloc d'instructions.

### F1 : Poser une question à Python

Quand nous écrivons `temperature > 20`, nous lui demandons si la valeur de `temperature` est strictement supérieure à 20. Cette **comparaison** donne une réponse de type **bool**, aussi appelée un **booléen** : `True` si elle est vraie, ou `False` si elle est fausse.

Prenez un moment pour prévoir les trois réponses avant de lancer la cellule, spécifiquement pour `20 > 20`.

**Attention aussi aux signes :** `=` sert à associer une valeur à un nom, alors que `==` permet de demander si deux valeurs sont égales.

````python
temperature = 21
print(temperature > 20)
print(20 > 20)
print(20 == 20)
````

### F2 : Que faire selon la réponse ?

Avec `if`, nous pouvons demander à Python d'exécuter un bloc seulement si une condition est vraie. Le mot `else` introduit ce qu'il doit faire dans l'autre cas.

Lisez l'exemple comme une phrase : « si la température est strictement supérieure à 20, affiche ce premier message ; sinon, affiche le second ».

Les deux-points annoncent un bloc, et les espaces au début des lignes indiquent quelles instructions en font partie. C'est ce qu'on appelle l'**indentation** : elle a un rôle dans le programme, ce n'est pas seulement une question de présentation.

Quelle branche sera exécutée pour 21 ? Et pour 20 ? Essayez de répondre avant de lancer la cellule.

````python
temperature = 21

if temperature > 20:
    print("Au-dessus du seuil")
else:
    print("Sous le seuil ou égal")
````

### F3 : Répéter une action

Une boucle `for` permet de parcourir les éléments d'une collection, ici notre liste de températures. Vous pouvez lire la première ligne comme ceci : « pour chaque température de cette liste, fais les instructions du bloc qui suit ».

À chaque **tour de boucle**, le nom `temperature` désigne la valeur suivante dans la liste. Python exécute le `print` avec cette valeur, puis passe au tour suivant.

````python
temperatures = [18.5, 21.0, 19.5]

for temperature in temperatures:
    print(temperature)
````

### F4 : Répéter tant que nécessaire

Avec `while`, nous ne demandons pas de parcourir une liste. Nous demandons de répéter un bloc **tant qu'une condition reste vraie**. Python vérifie cette condition avant chaque tour, y compris le tout premier.

Dans cet exemple, `compteur` vaut 0 au départ. À chaque tour, nous affichons sa valeur, puis nous lui ajoutons 1. À un moment, la condition `compteur < 3` devient fausse et la boucle s'arrête.

Essayez de retrouver les valeurs affichées avant d'exécuter. Pourquoi la valeur 3 ne sera-t-elle pas affichée ? Et quelle ligne permet justement à la condition de finir par devenir fausse ?

````python
compteur = 0

while compteur < 3:
    print(compteur)
    compteur = compteur + 1
````

Prenez un instant pour repérer **Kernel / Noyau → Interrupt / Interrompre**, ou le bouton d'arrêt de votre interface. Si un calcul reste bloqué, cette commande permet de l'arrêter.

Nous n'allons pas faire tourner exprès une boucle infinie. Pour la comprendre, un petit tableau de suivi suffit !

## H : Que faire quand Python nous signale un problème ?

Un message d'erreur peut sembler impressionnant au début, surtout lorsqu'il occupe plusieurs lignes. Il nous donne pourtant des indices pour comprendre à quel endroit Python n'a pas pu poursuivre.

Nous allons prendre un exemple ensemble, puis vérifier que notre carnet fonctionne encore lorsque nous repartons avec un noyau neuf. L'objectif n'est pas de ne jamais faire d'erreur, mais de savoir par où commencer pour en chercher la cause.

### H1 : Chercher ce qui ne convient pas

Regardons cette instruction :

```python
print("Température : " + 18.5)
```

Nous souhaitons afficher un texte suivi d'un nombre. Mais en utilisant `+`, nous demandons d'abord à Python de combiner directement ces deux objets. Comme il s'agit ici d'un texte et d'un nombre, cette opération provoque un **TypeError**, c'est-à-dire une erreur liée aux types utilisés.

Une possibilité consiste à fournir le texte et le nombre comme deux arguments distincts à `print()`, en les séparant par une virgule :

```python
print("Température :", 18.5)
```

Nous demandons alors à `print()` d'afficher ces deux éléments, sans essayer de les additionner auparavant. Regardez les exemples proposés sur la slide, elle vous aidera à aborder les prochains messages d'erreur.
