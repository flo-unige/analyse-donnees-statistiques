## E: How can we reuse an operation without rewriting it?

When we write programs, some tasks come up repeatedly: counting items, rounding a number or converting a unit. A **function** groups the instructions needed for a task and gives them a name so that we can reuse them.

We will start with functions that Python already provides. Then we will look at how to define our own function for a simple conversion. The main aim is to understand how to use and adapt it, rather than to memorise everything.


### E1: Asking a function for a result

Think of a **function call** as asking a tool to do something. For example, `len(temperatures)` means "count the items in this list". The name before the parentheses tells Python which tool to use, and the values inside the parentheses are the **arguments** you give it.

Here, `len()` counts the items, `max()` finds the largest value and `round()` rounds a number. In `round(18.567, ndigits=1)`, the second argument specifies that we want one digit after the decimal point.

Try to predict the three results before running the cell. Also look at the two lines involving `maximum`: the first stores the result under that name, while the second displays it.


````python
temperatures = [18.5, 21.0, 19.5]
print(len(temperatures))

maximum = max(temperatures)
print(maximum)

print(round(18.567, ndigits=1))
````

## F: How do we control what our program does?

So far, our instructions have simply run one after another. But sometimes we need to choose an action based on a result, or repeat the same operation for several measurements.

We will now explore **conditions**, which let us choose which instructions to execute, and **loops**, which let us repeat a block of instructions.


### F1: Asking Python a question

When we write `temperature > 20`, we are asking whether the value of `temperature` is strictly greater than 20. This **comparison** produces a value of type **bool**, also called a **Boolean**: `True` if the statement is true, or `False` if it is false.

Take a moment to predict the three answers before running the cell, paying particular attention to `20 > 20`.

**Watch the symbols:** `=` associates a value with a name, whereas `==` asks whether two values are equal.


````python
temperature = 21
print(temperature > 20)
print(20 > 20)
print(20 == 20)
````

### F2: What should happen next?

With `if`, we can ask Python to run a block only when a condition is true. The keyword `else` introduces what it should do otherwise.

Read the example as a sentence: "If the temperature is strictly greater than 20, display the first message; otherwise, display the second."

The colon introduces a block, and the spaces at the start of the lines show which instructions belong to it. This is called **indentation**: it affects how the program works, rather than just how it looks.

Which branch will run for 21? What about 20? Try to answer before running the cell.


````python
temperature = 21

if temperature > 20:
    print("Above the threshold")
else:
    print("At or below the threshold")
````

### F3: Repeating an action

A `for` loop goes through the items in a collection : here, our list of temperatures. You can read the first line like this: "For each temperature in this list, carry out the instructions in the block below."

On each **iteration**, or pass through the loop, the name `temperature` refers to the next value in the list. Python executes `print` with that value, then moves on to the next iteration.


````python
temperatures = [18.5, 21.0, 19.5]

for temperature in temperatures:
    print(temperature)
````

### F4: Repeating as long as needed

With `while`, we are not asking Python to go through a list. We are asking it to repeat a block **as long as a condition remains true**. Python checks the condition before every iteration, including the very first one.

In this example, `counter` starts at 0. On each iteration, we display its value, then add 1 to it. Eventually, the condition `counter < 3` becomes false and the loop stops.

Try to work out which values will be displayed before running the cell. Why will 3 not be displayed? Which line allows the condition to eventually become false?


````python
counter = 0

while counter < 3:
    print(counter)
    counter = counter + 1
````

## H: What should we do when Python reports a problem?

An error message can look intimidating at first, especially when it spans several lines. However, it gives us clues about where Python was unable to continue.

We will look at an example together, then check that our notebook still works when we start with a fresh kernel. The aim is not to avoid every error, but to know where to start looking for its cause.


### H1: Finding out what went wrong

Look at this instruction:

```python
print("Temperature: " + 18.5)
```

We want to display some text followed by a number. But by using `+`, we first ask Python to combine these two objects directly. Since one is text and the other is a number, this operation raises a **TypeError** : an error related to the types being used.

One option is to pass the text and the number to `print()` as two separate arguments, with a comma between them:

```python
print("Temperature:", 18.5)
```

We are then asking `print()` to display both items, without trying to add them together first. Look at the examples on the slide: they will help you work through the next error messages.
