## F. Modèle normal

### Construire une courbe à l'échelle de sa variable

Une **loi normale** est un modèle symétrique en forme de cloche. Sa moyenne `mu` fixe son centre et son écart-type `sigma` son étalement.

Nous allons utiliser vos résumés descriptifs pour construire une courbe à la même échelle. Cela ne veut pas dire que notre variable suit une loi normale, une conclusion « ce modèle convient mal » est tout à fait possible.

Pour construire cette courbe, il nous faut au moins deux mesures et un écart-type strictement positif. Si votre série est constante, revenez au choix de variable avec moi.

---
